# Funding source

```python
funding_sources = []

for _, row in table_disease.iterrows():
  list_sources = (str(row["Funding Details"])).split("; ")
  funding_sources = funding_sources + list_sources

split_comma = []

for source in funding_sources:
  split_comma = split_comma + source.split(", ")

# Many strings that will appear in the list are just the acronym of the funding sponsor or the grant number
split_comma
```

```python
from collections import Counter
ranking = Counter(split_comma).most_common(30)
ranking
```

# Country

```python
countries_per_paper = []

for _, row in table_disease.iterrows():

  affiliation_list = (str(row["Affiliations"])).split("; ")
  countries_per_affiliation = []

  for institution in affiliation_list:

    country_name = (institution.split(", "))[-1]
    countries_per_affiliation.append(country_name)

  countries_per_affiliation = list(set(countries_per_affiliation))
  countries_per_paper = countries_per_paper + countries_per_affiliation

countries_per_paper
```

```python
from collections import Counter
ranking = Counter(countries_per_paper).most_common(10)
ranking
```

```python
full_list = Counter(countries_per_paper)
full_list
```
